```python
WANTED = ["CompanyName", "CompanyNumber", "RegAddress.AddressLine1", "RegAddress.PostTown",
          "RegAddress.PostCode", "CompanyCategory", "CompanyStatus", "IncorporationDate",
          "Accounts.NextDueDate", "Accounts.AccountCategory", "SICCode.SicText_1",
          "ConfStmtNextDueDate"]

CATEGORY_COLS = ["CompanyCategory", "CompanyStatus", "Accounts.AccountCategory",
                 "RegAddress.PostTown", "SICCode.SicText_1"]

DATE_COLS = ["IncorporationDate", "Accounts.NextDueDate", "ConfStmtNextDueDate"]

read_kwargs = dict(
    usecols=[raw[c] for c in WANTED],
    dtype={**{raw[c]: "category" for c in CATEGORY_COLS},
           raw["CompanyNumber"]: "string"},
    parse_dates=[raw[c] for c in DATE_COLS],
    date_format="%d/%m/%Y",
    on_bad_lines="warn",
)
```

```python
missing_due = df2["Accounts.NextDueDate"].isna()
print(f"{missing_due.sum():,} companies have no accounts due date\n")

print(df2.loc[missing_due, "CompanyCategory"].value_counts().head(8))
print()
print(df2.loc[missing_due, "Accounts.AccountCategory"].value_counts().head(8))

#deciding factor of population
#we will now select Only analyse organisations that are supposed to file accounts at Companies House.
```

```python
df2["CompanyCategory"].value_counts()
```

```python
has_due = df2.groupby("CompanyCategory", observed=True)["Accounts.NextDueDate"].apply(lambda s: s.notna().mean())
has_due.sort_values(ascending=False).round(3)
```

```python
#which companies belong in our overdue analysis, defined by a clear, data-based rule?
IN_SCOPE_TYPES = has_due[has_due >= 0.9].index.tolist()
print(len(IN_SCOPE_TYPES), "types in scope:")
for t in IN_SCOPE_TYPES:
    print("  -", t)

df2["in_scope"] = df2["CompanyCategory"].isin(IN_SCOPE_TYPES)

print(f"\nIn scope: {df2['in_scope'].sum():,} companies ({df2['in_scope'].mean():.1%} of the register)")
print(f"Missing accounts due date INSIDE scope:  {df2.loc[df2['in_scope'], 'Accounts.NextDueDate'].isna().mean():.3%}")
print(f"Missing accounts due date OUTSIDE scope: {df2.loc[~df2['in_scope'], 'Accounts.NextDueDate'].isna().mean():.1%}")
```

```python
#how old is each company on the snapshot date?
df2["company_age_years"] = (snapshot - df2["IncorporationDate"]).dt.days / 365.25

print(df2["company_age_years"].describe().round(2))
print(f"\nNegative ages: {(df2['company_age_years'] < 0).sum():,}")
print(f"Older than 150 years: {(df2['company_age_years'] > 150).sum():,}")
```

```python
#are the 150+ year-old companies real, or data errors?
old = df2[df2["company_age_years"] > 150]

print(old["CompanyCategory"].value_counts().head(6))
print()
print(old.nlargest(8, "company_age_years")[["CompanyName", "CompanyCategory", "IncorporationDate", "company_age_years"]])
print()
print(f"Old companies inside our scope: {old['in_scope'].sum()} of {len(old)}")
```

```python
df2["accounts_overdue"] = df2["Accounts.NextDueDate"] < snapshot
df2["confstmt_overdue"] = df2["ConfStmtNextDueDate"] < snapshot

scope = df2[df2["in_scope"]]
print(f"Accounts overdue (in scope):               {scope['accounts_overdue'].mean():.2%}")
print(f"Confirmation statement overdue (in scope): {scope['confstmt_overdue'].mean():.2%}")
print(f"Both overdue (in scope):                   {(scope['accounts_overdue'] & scope['confstmt_overdue']).mean():.2%}")
print()
print(pd.crosstab(scope["accounts_overdue"], scope["confstmt_overdue"], margins=True))
```

```python
by_status = scope.groupby("CompanyStatus", observed=True).agg(
    companies=("CompanyNumber", "size"),
    accounts_overdue_rate=("accounts_overdue", "mean"),
    confstmt_overdue_rate=("confstmt_overdue", "mean"),
).sort_values("companies", ascending=False)

by_status.round(3)
```

```python
df2["is_active"] = df2["CompanyStatus"].eq("Active")
df2["strike_off_proposal"] = df2["CompanyStatus"].eq("Active - Proposal to Strike off")

active = df2[df2["in_scope"] & df2["is_active"]]
scope = df2[df2["in_scope"]]

print(f"Active in-scope companies: {len(active):,}")
print(f"  Outcome 1 baseline: accounts overdue  = {active['accounts_overdue'].mean():.2%}")
print(f"  (also) confirmation statement overdue = {active['confstmt_overdue'].mean():.2%}")
print()
print(f"All in-scope companies: {len(scope):,}")
print(f"  Outcome 2 baseline: proposal to strike off = {scope['strike_off_proposal'].mean():.2%}")
```

```python
#what kind of business does each company claim to do, grouped into broad industries?
df2["sic_code"] = df2["SICCode.SicText_1"].astype("string").str.extract(r"^(\d{5})", expand=False)

division = pd.to_numeric(df2["sic_code"].str[:2], errors="coerce")

bins   = [0, 3, 9, 33, 35, 39, 43, 47, 53, 56, 63, 66, 68, 75, 82, 84, 85, 88, 93, 96, 98, 99]
labels = list("ABCDEFGHIJKLMNOPQRSTU")
df2["sic_section"] = pd.cut(division, bins=bins, labels=labels)

df2["sic_section"] = df2["sic_section"].cat.add_categories(["Dormant", "None supplied"])
df2.loc[df2["sic_code"] == "99999", "sic_section"] = "Dormant"
df2.loc[df2["sic_code"].isna(), "sic_section"] = "None supplied"

df2["sic_section"].value_counts(normalize=True).head(10).round(3)
```

```python
df2["sic_section"].value_counts(normalize=True)[["Dormant", "None supplied"]].round(4)
```

```python
#which companies are registered in Glasgow?
pc = (df2["RegAddress.PostCode"].astype("string")
        .str.upper()
        .str.replace(r"\s+", "", regex=True))

df2["postcode_clean"] = pc
df2["postcode_area"] = pc.str.extract(r"^([A-Z]{1,2})\d", expand=False)
df2["glasgow"] = df2["postcode_area"].eq("G").fillna(False).astype(bool)

print(f"Glasgow (G area): {df2['glasgow'].sum():,} entities ({df2['glasgow'].mean():.2%})")
print(df2.loc[df2["postcode_area"].isin(["G", "GL", "GU"]), "postcode_area"].value_counts())
```
